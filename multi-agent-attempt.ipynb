```python
import math
from typing import Optional, Tuple
from smolagents import (
    tool, CodeAgent, TransformersModel,
    DuckDuckGoSearchTool, VisitWebpageTool,
)

@tool
def calculate_cargo_travel_time(
    origin_coords: Tuple[float, float],
    destination_coords: Tuple[float, float],
    cruising_speed_kmh: Optional[float] = 750.0,
) -> float:
    """
    Estimate cargo plane travel time between two (lat, lon) points.

    Args:
        origin_coords: (latitude, longitude) of the start.
        destination_coords: (latitude, longitude) of the destination.
        cruising_speed_kmh: Cruising speed in km/h. Defaults to 750.

    Returns:
        Estimated travel time in hours.
    """
    lat1, lon1 = map(math.radians, origin_coords)
    lat2, lon2 = map(math.radians, destination_coords)
    a = (math.sin((lat2 - lat1) / 2) ** 2
         + math.cos(lat1) * math.cos(lat2) * math.sin((lon2 - lon1) / 2) ** 2)
    distance_km = 6371.0 * 2 * math.asin(math.sqrt(a)) * 1.1  # +10% detour
    return round(distance_km / cruising_speed_kmh + 1.0, 2)  # +1h takeoff/landing
```

```python
model = TransformersModel(
    model_id="Qwen/Qwen2.5-Coder-1.5B-Instruct",
    device_map="auto",
    max_new_tokens=2048,
)

web_agent = CodeAgent(
    model=model,
    tools=[DuckDuckGoSearchTool(), VisitWebpageTool()],
    name="web_agent",
    description="Browses the web to find information",
    max_steps=10,
    verbosity_level=0,
)

manager_agent = CodeAgent(
    model=model,
    tools=[calculate_cargo_travel_time],
    managed_agents=[web_agent],
    additional_authorized_imports=["plotly", "geopandas", "shapely", "json", "pandas", "numpy"],
    planning_interval=5,
    max_steps=15,
)
```
